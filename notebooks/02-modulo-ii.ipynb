```python
nomes = np.array(['Bob', 'Joe', 'Will', 'Bob', 'Will', 'Joe', 'Joe'])
print(nomes)
dados = rng.standard_normal((7, 4))
print(nomes == 'Bob') #Filtro que será aplicado
print(dados[nomes == 'Bob']) #Linhas contendo os números associados ao Bob
#Exibindo apenas os registros de dados que não possuem Bob como nome associado
print(nomes != 'Bob')
print(dados[~(nomes == 'Bob')])
#Criando uma variável para armazenar o resultado do filtro e então aplicá-lo como indexador
condicao = nomes == 'Will'
print(dados[condicao])
condicao = nomes == 'Bob'
print(dados[~condicao])
```

### Funções Universais

```python
matriz1 = rng.standard_normal((8, 3))
matriz2 = rng.standard_normal((8, 3))
print(np.add(matriz1, matriz2)) #Soma elemento a elemento das matrizes
print(np.matmul(matriz1, matriz2.T)) #Realiza multiplicação de matrizes
print(np.negative(matriz1)) #Multiplica todos os elementos por -1
print(np.mod(matriz1, matriz2)) #Retorna o resto da divisão de elemento por elemento
print(np.absolute(matriz1)) #Retorna o absoluto de cada elemento
print(np.rint(matriz1)) #Arredonda cada elemento para o inteiro mais próximo
```

```python
print(np.sin(matriz1)) #Calcula o seno de cada elemento
print(np.cos(matriz1)) #Calcula o cosseno de cada elemento
print(np.tan(matriz1)) #Calcula a tangente de cada elemento
```

```python
#Retorna True nos elementos correspondente a matriz1 > matriz2
print(np.greater(matriz1, matriz2))
#Retorna True nos elementos correspondente a matriz1 >= matriz2
print(np.greater_equal(matriz1, matriz2))
#Retorna True nos elementos correspondente a matriz1 < matriz2
print(np.less(matriz1, matriz2))
#Retorna True nos elementos correspondente a matriz1 <= matriz2
print(np.less_equal(matriz1, matriz2))
#Retorna True nos elementos correspondente a matriz1 != matriz2
print(np.not_equal(matriz1, matriz2))
#Retorna True nos elementos correspondente a matriz1 == matriz2
print(np.equal(matriz1, matriz2))
```

### Valores ausentes (`NaN`) e máscaras booleanas

```python
t = np.array([21.5, np.nan, 23.0, 19.5, np.nan, 25.0])
print(t.mean(), np.nanmean(t), np.nanstd(t), np.nanmax(t))
print(np.isnan(t), np.isnan(t).sum())
print(t == np.nan, np.nan == np.nan)
```

```python
valido = t[~np.isnan(t)]              # descarta os ausentes
print(valido)
t2 = t.copy()                         # copia para não alterar o original
t2[np.isnan(t2)] = np.nanmean(t)      # imputa com a média dos válidos
print(t2)
print(np.where(np.isnan(t), 0.0, t))  # alternativa funcional: condição, se_sim, se_nao
quente = t > 22                       # comparações com NaN dão False
print(quente, t[quente])
print(np.isnan(t).any(), np.count_nonzero(~np.isnan(t)))
```

### Álgebra linear básica

```python
A = np.array([[2., 1.], [1., 3.]])
B = np.array([[0., 1.], [1., 0.]])
print(A * B)                     # elemento a elemento
print(A @ B)                     # produto matricial
print(np.dot(A, B))
v1 = np.array([1., 2., 3.])
v2 = np.array([4., 5., 6.])
print(v1 @ v2, np.linalg.norm(v1))
```

```python
b = np.array([5., 10.])
x = np.linalg.solve(A, b)
print(x, np.allclose(A @ x, b))
A_inv = np.linalg.inv(A)
print(A_inv)
print(np.allclose(A @ A_inv, np.eye(2)))
print(round(np.linalg.det(A), 6), np.trace(A))
```

```python
valores, vetores = np.linalg.eigh(A)
print(valores)
print(np.allclose(A @ vetores[:, 0], valores[0] * vetores[:, 0]))
```

### Visões, cópias e ordenação indireta

```python
a = np.arange(12).reshape(3, 4)
print(np.shares_memory(a, a.reshape(4, 3)))   # visão
print(np.shares_memory(a, a[:, 1:3]))         # fatia: visão
print(np.shares_memory(a, a[[0, 2]]))         # lista de índices: cópia
print(np.shares_memory(a, a[a > 5]))          # máscara booleana: cópia
```

```python
notas = np.array([7.5, 9.0, 6.0, 8.5, 5.5])
ordem = np.argsort(notas)
print(ordem, notas[ordem])
print(notas[ordem[::-1][:3]])                 # as três maiores notas
print(np.argmax(notas), np.where(notas >= 7))
print(np.cumsum(notas), np.diff(notas))       # acumulada e diferenças consecutivas
print(np.clip(notas, 6, 9))                   # limita os valores ao intervalo [6, 9]
```

### Vetorização versus laços

```python
import timeit
dados_v = np.random.default_rng(1).random(100_000)

def com_laco():
    total = 0.0
    for valor in dados_v:
        total += valor * valor
    return total

def vetorizado():
    return float(dados_v @ dados_v)

print(np.isclose(com_laco(), vetorizado()))
t_laco = timeit.timeit(com_laco, number=5)
t_vet = timeit.timeit(vetorizado, number=5)
print(f"laço: {t_laco:.4f} s | vetorizado: {t_vet:.4f} s | razão: {t_laco / t_vet:.0f}x")
```
