### Resolução — Módulo VIII (Transformers e LLMs)

```python
import numpy as np

def softmax(x, axis=-1):
    x = x - x.max(axis=axis, keepdims=True)   # estabilidade numérica
    e = np.exp(x)
    return e / e.sum(axis=axis, keepdims=True)

def attention(Q, K, V):
    dk = Q.shape[-1]
    scores = Q @ K.T / np.sqrt(dk)            # (n, n)
    pesos = softmax(scores, axis=-1)          # linhas somam 1
    return pesos @ V, pesos

rng = np.random.default_rng(0)               # API moderna (NumPy 2.x)
Q = rng.random((3, 4)); K = rng.random((3, 4)); V = rng.random((3, 4))
saida, pesos = attention(Q, K, V)
assert np.allclose(pesos.sum(axis=1), 1.0)   # verifica que cada linha soma 1
print(saida.shape)                            # (3, 4)
```

### Resolução — Módulo X (Agentes Inteligentes)

```python
import re
def calculadora_segura(expr):
    if not re.fullmatch(r"[0-9+\-*/(). ]+", expr):   # allowlist
        return "erro: expressão não permitida"
    return str(eval(expr, {"__builtins__": {}}))
```

```python
import re
CPF = re.compile(r'\b\d{3}\.?\d{3}\.?\d{3}-?\d{2}\b')
def guardrail_saida(texto):
    if CPF.search(texto):
        return "[BLOQUEADO: possível dado sensível na resposta]"
    return texto

print(guardrail_saida("O CPF é 123.456.789-00"))  # -> bloqueado
```
