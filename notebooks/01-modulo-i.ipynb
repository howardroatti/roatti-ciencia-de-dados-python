```python
def menorIdade(idade1 = 0, idade2 = 0, idade3 = 0):
    """Esse método irá verificar qual a menor entre três idades informadas"""
    if idade1 == idade2 and idade2 == idade3:
        return 0
    elif idade1 > idade2 and idade2 > idade3:
        return idade3
    elif idade2 > idade3 and idade3 > idade1:
        return idade1
    else:
        return idade2
print(menorIdade(idade1 = 10, idade2 = 11, idade3 = 18))
```

#### Pacotes

```python
# Fibonacci numbers module
def fib(n): # write Fibonacci series up to n
    a, b = 0, 1
    while a < n:
        print(a, end=' ')
        a, b = b, a+b
    print()
def fib2(n): # return Fibonacci series up to n
    result = []
    a, b = 0, 1
    while a < n:
        result.append(a)
        a, b = b, a+b
    return result
```

```python
import sys, types
fibonacci = types.ModuleType("fibonacci")
fibonacci.fib, fibonacci.fib2 = fib, fib2
sys.modules["fibonacci"] = fibonacci
```

```python
from urllib.request import urlopen
from urllib.error import URLError
import io

# Cópias locais (simplificadas) das páginas usadas nos exemplos
HTML_OFFLINE = {
    "aphrodite": '<html>\n<head>\n<title>Profile: Aphrodite</title>\n</head>\n<body bgcolor="yellow">\n<center>\n<br><br>\n<img src="/static/aphrodite.gif" />\n<h2>Name: Aphrodite</h2>\n<br><br>\nFavorite animal: Dove\n<br><br>\nFavorite color: Red\n<br><br>\nHometown: Mount Olympus\n</center>\n</body>\n</html>\n',
    "poseidon": '<html>\n<head>\n<title >Profile: Poseidon</title>\n</head>\n<body bgcolor="yellow">\n<center>\n<br><br>\n<img src="/static/poseidon.jpg" />\n<h2>Name: Poseidon</h2>\n<br><br>\nFavorite animal: Dolphin\n<br><br>\nFavorite color: Blue\n<br><br>\nHometown: Sea\n</center>\n</body>\n</html>\n',
    "dionysus": '<html>\n<head>\n<TITLE >Profile: Dionysus</title  / >\n</head>\n<body bgcolor="yellow">\n<center>\n<br><br>\n<img src="/static/dionysus.jpg" />\n<h2>Name: Dionysus</h2>\n<img src="/static/grapes.png"><br><br>\nHometown: Mount Olympus\n<br><br>\nFavorite animal: Leopard <br>\n<br>\nFavorite Color: Wine\n</center>\n</body>\n</html>\n',
}

def abrir(url):
    """urlopen(url); sem internet, devolve a cópia local da página."""
    try:
        return urlopen(url, timeout=10)
    except (URLError, OSError):
        return io.BytesIO(HTML_OFFLINE[url.rsplit("/", 1)[1]].encode("utf-8"))

url = "http://olympus.realpython.org/profiles/aphrodite"
page = abrir(url)
page
```

```python
import re
url = "http://olympus.realpython.org/profiles/dionysus"
page = abrir(url)
html = page.read().decode("utf-8")
pattern = "<title.*?>.*?</title.*?>"
match_results = re.search(pattern, html, re.IGNORECASE)
title = match_results.group()
title = re.sub("<.*?>", "", title) # Remove as tags HTML
print(title)
```
