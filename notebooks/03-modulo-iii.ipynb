### Agrupamento avançado: `agg`, `transform` e `filter`

```python
v = vendas.merge(produtos, on="id_produto")
v["receita"] = v["quantidade"] * v["preco"] * (1 - v["desconto"].fillna(0))
resumo = v.groupby("categoria").agg(
    vendas=("id_venda", "size"),
    unidades=("quantidade", "sum"),
    receita_total=("receita", "sum"),
    ticket_medio=("receita", "mean"),
)
print(resumo.round(2))
```

```python
print(v.groupby(["categoria", "canal"])["receita"].sum().round(2))
print(v.groupby("categoria")["receita"].agg(["min", "median", "max"]).round(2))
print(v.groupby("canal", as_index=False)["receita"].sum().round(2))
```

```python
v["receita_pct_cat"] = v["receita"] / v.groupby("categoria")["receita"].transform("sum")
print(v[["categoria", "receita", "receita_pct_cat"]].head())
print(v.groupby("categoria")["receita_pct_cat"].sum())
v["receita_z"] = v.groupby("categoria")["receita"].transform(lambda x: (x - x.mean()) / x.std())
print(v.groupby("categoria")["receita_z"].agg(["mean", "std"]).round(6))
```

### Encadeamento de operações e boas práticas

```python
analise = (
    passageiros
    .dropna(subset=["Age"])
    .assign(faixa=lambda d: pd.cut(d["Age"], [0, 18, 60, 100],
                                   labels=["jovem", "adulto", "idoso"]))
    .query("Fare > 10")
    .groupby(["Pclass", "faixa"], observed=True)
    .agg(n=("PassengerId", "size"), tarifa=("Fare", "mean"), sobrev=("Survived", "mean"))
    .round(2)
    .reset_index()
)
print(analise)
```

```python
import timeit
teste = pd.DataFrame({"a": np.arange(2000), "b": np.arange(2000)[::-1]})

def com_laco():
    saida = []
    for _, linha in teste.iterrows():
        saida.append(linha["a"] * linha["b"])
    return saida

def vetorizado():
    return teste["a"] * teste["b"]

print(np.array_equal(com_laco(), vetorizado().to_numpy()))
print(f"laço: {timeit.timeit(com_laco, number=3):.3f} s | "
      f"vetorizado: {timeit.timeit(vetorizado, number=3):.4f} s")
```
